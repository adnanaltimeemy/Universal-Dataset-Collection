```python
df = pd.read_csv("food_tare.csv")

# Preview data
df.head()
```

```python
df.info()
df.describe()

# Check missing values
df.isnull().sum()
```

```python
df = df.ffill(limit=2)
```

```python
# Drop duplicates
df = df.drop_duplicates()

# Fill missing values (forward fill)
df = df.ffill()
```

```python
le = LabelEncoder()

for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = le.fit_transform(df[col])
```
