```python
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Preprocessing
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# NLP
from sklearn.feature_extraction.text import TfidfVectorizer

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('Customer_Segmentation.csv')
df.head()
```

```python
data = {
    'CustomerID': [1,2,3,4,5],
    'Age': [25, 45, 35, 23, 52],
    'Annual_Income': [40000, 80000, 60000, 20000, 90000],
    'Spending_Score': [60, 30, 50, 80, 20],
    'Review': [
        "I love this product it is amazing",
        "Not satisfied poor quality",
        "Good value for money",
        "Excellent service and fast delivery",
        "Very bad experience will not buy again"
    ]
}

df = pd.DataFrame(data)
```

```python
# Check nulls
df.isnull().sum()

# Fill missing text
df['Review'] = df['Review'].fillna("")
```

```python
vectorizer = TfidfVectorizer(stop_words='english')

text_features = vectorizer.fit_transform(df['Review'])

text_df = pd.DataFrame(
    text_features.toarray(),
    columns=vectorizer.get_feature_names_out()
)
```

```python
numerical_features = df[['Age', 'Annual_Income', 'Spending_Score']]

final_df = pd.concat([numerical_features.reset_index(drop=True), text_df], axis=1)
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(final_df)
```

```python
n_samples = scaled_data.shape[0]

for i in range(1, n_samples + 1):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(scaled_data)
    wcss.append(kmeans.inertia_)
```
